### 2. Implement a Resilience Mechanism (Idempotency or Retries)

Under high concurrency, network glitches or transient database locks happen. Demonstrating failure handling makes your code production-ready.

* **In `order-service`:** Add an `Idempotency-Key` header check using a unique UUID to prevent duplicate order writes if a network retry occurs during load testing.
* **In `cdc-worker`:** Implement a basic exponential backoff retry loop when inserting events into `olap-db` to handle temporary database connection drops.
